```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from mlp.naive_mlp import NMLP
```

```python
df_train = pd.read_parquet("data/mnist_train.parquet")
df_test = pd.read_parquet("data/mnist_test.parquet")
```

```python
df_X_train = df_train.drop("y", axis=1)
df_Y_train = pd.get_dummies(df_train["y"], dtype="int")

df_X_test = df_test.drop("y", axis=1)
df_Y_test = pd.get_dummies(df_test["y"], dtype="int")
```

```python
plt.imshow(df_X_train.iloc[0].to_numpy().reshape(28,28), cmap="gray")
```

```python
df_train.to_numpy()
```

```python
layer_sizes = (784,30, 10)

X_train = df_X_train.to_numpy().reshape(60000,784,1)
Y_train = df_Y_train.to_numpy().reshape(60000,10,1)

X_test = df_X_test.to_numpy().reshape(10000,784,1)
Y_test = df_Y_test.to_numpy().reshape(10000,10,1)

mlp = NMLP(layer_sizes)

mlp.fit(X_train, Y_train, 1, 30, X_test, Y_test)

import matplotlib.pyplot as plt

plt.plot(mlp.training_loss)
plt.plot(mlp.test_loss)
plt.show()
```

```python
plt.plot(mlp.training_loss)
plt.plot(mlp.test_loss)
plt.show()
```

```python
true = []
pred = []

for x, y in zip(X_test, Y_test):
    true.append(np.argmax(y))
    pred.append(np.argmax(mlp.predict(x)))
```

```python
print(np.mean([p==t for p,t in zip(pred, true)]))
```

```python
for i in range(len(pred)):
    if (pred[i] != true[i]):
        print(i, pred[i], true[i])
```
